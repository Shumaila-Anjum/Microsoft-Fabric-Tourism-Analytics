```python
# Welcome to your new notebook
# Type here in the cell editor to add code!
df = spark.read.table("tourism_Data")
display(df)
```

```python
from pyspark.sql.functions import col
df2 =   df.withColumn(
    "Revenue_Per_Tourist",
    col("Revenue")/col("Tourist_Visits")
)
display(df2)
```

```python
high_revenue = df2.filter(
    col("Revenue_Per_Tourist") >= 40
)
display(high_revenue)
```

```python
from pyspark.sql.functions import sum
region_revenue = df2.groupBy("Region").agg(sum("Revenue").alias("Total_Revenue")
)
display(region_revenue)
```

```python
from pyspark.sql.functions import sum
region_tourists_visit = df2.groupBy("Region").agg(sum("Tourist_Visits").alias("Total_Tourists_Visits")
)
display(region_tourists_visit)
```

```python
from pyspark.sql.functions import avg
region_satisfaction = df2.groupBy("Region").agg(
    avg("Customer_Satisfaction").alias("Avg_Satisfaction")
)
display(region_satisfaction)
```

```python
# Revenue by Region
from pyspark.sql.functions import sum
df.groupBy("Region").agg(sum("Revenue").alias("Total_Revenue")).orderBy("Total_Revenue",ascending=False).show()
```

```python
# Highest tourist region
from pyspark.sql.functions import sum
df.groupBy("Region").agg(sum("Tourist_Visits").alias("Total_Visitors")).orderBy("Total_Visitors",ascending=False).show()
```

```python
# Average customer satisfaction
from pyspark.sql.functions import avg
df.groupBy("Region").agg(avg("Customer_Satisfaction").alias("Avg_Customer_Satisfaction")).orderBy("Avg_Customer_Satisfaction",ascending=False).show()
```

```python
#Find the top revenue month
from pyspark.sql.functions import sum
df.groupBy("Month").agg(sum("Revenue").alias("Total_Revenue")).orderBy("Total_Revenue",ascending=False).show()
```
